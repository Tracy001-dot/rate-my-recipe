```python
import os
import pandas as pd

#set the path to the data folder
data_dir = '../data'

#checking files available in the data folder
print("files in the data folder:", os.listdir(data_dir))

#loading the recipe and interaction data
recipes_path = os.path.join(data_dir, 'RAW_recipes.csv')
interactions_path = os.path.join(data_dir, 'RAW_interactions.csv')

recipes_df = pd.read_csv(recipes_path)
interactions_df = pd.read_csv(interactions_path)

#displaying the size of each dataset
print("recipes dataset shape:", recipes_df.shape)
print("interactions dataset shape:", interactions_df.shape)

#preview the first 3 rows of the ricpe dataset
display(recipes_df.head(3))
```

```python
#check the number of rows and columns  in both data set

print("DATASET SHAPES")
print("Recipes dataset:", recipes_df.shape)
print("interaction dataset:",interactions_df.shape)

#display information about the recipe dataset
print("\nRECIPES DATASET INFO")
recipes_df.info()

#preview the first three (3) row of the interaction dataset
print("\nINTERACTIONS DATASET PREVIEW")
display(interactions_df.head(3))
```

```python
#display information about the interaction dataset
print("INTERACTION DATASET INFO")
interactions_df.info()

#check how often each rating appears
print("\n RATING DISTRIBUTION")
print(interactions_df['rating'].value_counts().sort_index())

```

```python
#calculate the average rating and total review count for each recipe
recipe_stats = interactions_df[interactions_df['rating'] > 0].groupby('recipe_id').agg(
    avg_rating = ('rating', 'mean'),
    review_count = ('rating', 'count')
).reset_index()

#medge these stats back into our main recipes dataframe 
merged_recipes_df = recipes_df.merge(recipe_stats, left_on = 'id', right_on='recipe_id', how = 'inner')

#preview the medged datasets
print("medged dataset shape:", merged_recipes_df.shape)
display(merged_recipes_df[['name','minutes', 'avg_rating','review_count', 'n_ingredients']].head(5))
```

```python
#save the medged dataset to a CSV file
output_path = '../data/processed_recipes.csv'
merged_recipes_df.to_csv(output_path, index = False)

#confirmimg if the file has been saved
print(f"processed datset saved successfully to: {output_path}")
```

```python
#display the top recipes based on average rating
print("TOP RECIPES BY AVERAGE RATING")
top_naive = merged_recipes_df.sort_values(
    by = 'avg_rating',
    ascending=False
)
display(top_naive[['name','avg_rating', 'review_count']].head(5))

#display the recipe with the highest number of reviews
print("TOP RECIPES BY NUMBER OF REVIEWS")
top_reviewed = merged_recipes_df.sort_values(
    by = 'review_count',
    ascending=False
)
display(top_reviewed[['name', 'avg_rating','review_count']].head(5))
```

```python
import numpy as np

#calculating the global mean rating across all valid user reviews
global_mean = interactions_df[interactions_df['rating'] >0]['rating'].mean()
print(f"overall average rating:{global_mean:3f}")

#set confidence value
c = 10

#calculate the bayesian average for each receipt
def calculate_bayesian_average(row):
    n = row['review_count']
    r = row['avg_rating']
    return (c * global_mean + n * r) /( c + n)

#add the bayesian rating to the dataset
merged_recipes_df['bayesian_rating'] = merged_recipes_df.apply( calculate_bayesian_average, axis = 1)

#displaying the top 10 using bayesian rating
print("\n TOP 10 RECIPES BY BAYESIAN RATING")
top_bayesian = merged_recipes_df.sort_values(
    by = 'bayesian_rating', ascending=False
)

display(
    top_bayesian[
       ['name', 'avg_rating', 'review_count', 'bayesian_rating'] 
    ].head(10)
)
```

```python
# save the updated dataset with bayesian ratings
output_path = os.path.join(data_dir, 'processed_recipes.csv')
merged_recipes_df.to_csv(output_path, index = False)

#by confirming if the dataset has been saved
print(f" updated dataset saved to : {output_path}")
```

```python
#RECOMMENDATION

def recommend_recipes(keyword, top_n=5):
#find the recipes that contains the search keyword
 matches = merged_recipes_df[
  merged_recipes_df['name'].str.contains(
   keyword, case = False, na=False
   )
]

 #sort the results by bayesian ratings
 best_matches = matches.sort_values(by='bayesian_rating',ascending=False)

 #return the top recipes with their main details 
 return best_matches[['name','minutes','n_steps','avg_rating','review_count','bayesian_rating']
 ].head(top_n)

```
